```python
import scipy.stats as stats
import pandas as pd

# 가상의 데이터 생성 (예시)
data = pd.DataFrame({
    'Category': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Numeric': [10, 20, 15, 25, 12, 18]
})

# 명목형 변수를 이진 인코딩으로 변환
data['Binary_Category'] = (data['Category'] == 'A').astype(int)

# pointbiserialr 함수로 포인트-비스레 상관계수 계산
correlation_coefficient, p_value = stats.pointbiserialr(data['Binary_Category'], data['Numeric'])

print("Point-biserial correlation coefficient:", correlation_coefficient)
print("P-value:", p_value)
```
